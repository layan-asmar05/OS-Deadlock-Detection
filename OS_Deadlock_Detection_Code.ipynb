#Operating Systems - Phase 2: Deadlock Detection (Graph Reduction)



```python
from google.colab import files
```

```python
def check_deadlock(num_p, num_r, avail, alloc, req):
# Array to track finished processes (initially all False)
    finish = [False] * num_p
    execution_sequence = []

    print("\n--- Execution Log ---")

    while True:
        progress_made = False

       # Iterate through all processes to find an unblocked process
        for i in range(num_p):
            if not finish[i]:
                # Check if the current process's request is <= available resources
                can_run = True
                for j in range(num_r):
                    if req[i][j] > avail[j]:
                        can_run = False
                        break

                # If the process can run safely
                if can_run:
                    print(f"[+] Process P{i} can run (Request <= Available).")

                    # Release the allocated resources back to the system
                    for j in range(num_r):
                        avail[j] += alloc[i][j]

                    finish[i] = True
                    execution_sequence.append(f"P{i}")
                    progress_made = True

                    print(f"    -> P{i} reduced. New Available: {avail}")
                    break # Break the loop and restart scanning with updated available resources
      # If we iterated through all processes and no progress was made, stop
        if not progress_made:
            break

    print("-" * 21)

    # Check if any process remains unreduced (Deadlock verification)
    deadlocked_processes = [i for i in range(num_p) if not finish[i]]

    if deadlocked_processes:
        print("\n[Result] Deadlock Detected!")
        print(f"Processes involved in Deadlock: {['P' + str(p) for p in deadlocked_processes]}")
        return True
    else:
        print("\n[Result] No Deadlock! All processes were completely reduced.")
        print(f"Execution Sequence: {' -> '.join(execution_sequence)}")
        return False


```

```python
def read_input_file(filename):
    try:
        with open(filename, 'r') as file:
            lines = file.readlines()

       # Clean up lines from whitespaces and empty lines
        lines = [line.strip() for line in lines if line.strip()]

        num_p, num_r = map(int, lines[0].split())
        avail = list(map(int, lines[1].split()))

        alloc = []
        for i in range(num_p):
            alloc.append(list(map(int, lines[2+i].split())))

        req = []
        for i in range(num_p):
            req.append(list(map(int, lines[2+num_p+i].split())))

        return num_p, num_r, avail, alloc, req

    except FileNotFoundError:
        print(f"Error: The file '{filename}' was not found. Please make sure it's in the same folder.")
        exit()
```
